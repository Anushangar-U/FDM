### Outlier Decision

Extreme sensor values were investigated before deciding whether to remove them.

The temperature sensors showed relatively stable ranges, although S2_Temp contained higher values reaching 29.00°C. These higher observations occurred across multiple consecutive records and were associated with an occupancy count of 3, suggesting that they may represent genuine environmental changes rather than isolated sensor errors.

Higher sound measurements were also observed during occupied periods. Similarly, higher CO₂ values occurred across consecutive observations while the room was occupied, showing a meaningful progression rather than isolated invalid values.

S5_CO2_Slope contained both positive and negative values. These were retained because the variable represents the rate of change in CO₂, where both increasing and decreasing values are meaningful.

Therefore, no observations were removed solely because they were statistically extreme. Removing these values could eliminate potentially useful information for occupancy prediction.

### Feature Selection Decision

The four sensors within each of the temperature, light, and sound groups were highly
correlated with each other and with their respective group averages. This is expected,
since all four sensors were placed in the same small room and largely capture the same
underlying signal.

To reduce redundancy and multicollinearity — which particularly affects distance-based
and linear models such as KNN and logistic regression — the raw per-sensor columns for
temperature, light, and sound are dropped in favor of the engineered `Avg_*` and
`*_Spread` features, which retain both the central tendency and the cross-sensor
variability. `S5_CO2`, `S5_CO2_Slope`, `S6_PIR`, and `S7_PIR` are kept as-is, since they
are not repeated across multiple sensors and each carries distinct information.

This reduced feature set will be used as the default for all models. If tree-based
ensembles (Random Forest, Gradient Boosting) underperform expectations during Stage 6,
the raw per-sensor columns can be reintroduced for those specific models as a follow-up
experiment, since trees are less sensitive to multicollinearity.

### Categorical Feature Engineering

`CO2_Level` is created as an interpretable Low/Medium/High feature using project-defined CO2 bands. `Sound_Level` is created from training-data tertiles of `Avg_Sound`, so the sound thresholds are learned without using the held-out test session. The readable categories are retained in the prepared dataset, while numeric codes are used for modelling.

```python
# Group CO2 and sound readings into low, medium and high levels.

# Use the CO2 limits chosen for this project.
data["CO2_Level"] = np.select(
    [
        data["S5_CO2"] < 800,
        data["S5_CO2"] <= 1000
    ],
    [
        "Low",
        "Medium"
    ],
    default="High"
)

# Keep 23 December aside for testing.
# Use the other sessions to work out the sound cutoffs.
TEST_DATE = "2017/12/23"

train_mask = data["Session_Date"] != TEST_DATE
test_mask = data["Session_Date"] == TEST_DATE

# Work out the sound cutoffs using only the training rows.
sound_low = data.loc[train_mask, "Avg_Sound"].quantile(1/3)
sound_high = data.loc[train_mask, "Avg_Sound"].quantile(2/3)

data["Sound_Level"] = pd.cut(
    data["Avg_Sound"],
    bins=[-np.inf, sound_low, sound_high, np.inf],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

print("Sound thresholds:")
print("Low <=", sound_low)
print("Medium <=", sound_high)

print("\nCO2 categories:")
print(data["CO2_Level"].value_counts())

print("\nSound categories:")
print(data["Sound_Level"].value_counts())
```

```python
# Turn low, medium and high into 0, 1 and 2.

level_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

data["CO2_Level_Code"] = (
    data["CO2_Level"]
    .map(level_mapping)
    .astype(int)
)

data["Sound_Level_Code"] = (
    data["Sound_Level"]
    .map(level_mapping)
    .astype(int)
)

data[
    [
        "S5_CO2",
        "CO2_Level",
        "CO2_Level_Code",
        "Avg_Sound",
        "Sound_Level",
        "Sound_Level_Code"
    ]
].head(10)
```

## Preprocessing & Feature Engineering Summary

**Missing values and duplicates:** No missing values or duplicate rows were found, consistent with the EDA findings; no imputation or deduplication was required.

**Temporal handling and leakage prevention:** `Date` and `Time` were combined into a `DateTime` field and the observations were sorted chronologically. A `Session_Date` field was retained only for splitting. Instead of a random row split, the complete session `2017/12/23` was held out as the test set and the remaining sessions were used for training. This reduces the risk of temporally adjacent, highly similar observations appearing in both training and testing data. All four occupancy classes (0, 1, 2, and 3) are represented in both sets.

**Outlier treatment:** Extreme sensor readings (elevated temperature, sound, and CO2) were investigated individually rather than removed automatically. They occurred in consecutive records and coincided with occupied periods, indicating genuine occupancy-related signal rather than obvious sensor error, so all observations were retained.

**Feature engineering:** Aggregate features (`Avg_Temp`, `Avg_Light`, `Avg_Sound`, `Temp_Spread`, `Light_Spread`, `PIR_Activity`) were created from the repeated per-sensor columns to summarize both central tendency and cross-sensor variability.

**Categorical feature engineering:** `CO2_Level` was created using project-defined Low/Medium/High CO2 bands. `Sound_Level` was created using tertile thresholds learned from the training data only. Both were ordinally encoded as Low=0, Medium=1, High=2 for modelling, while the readable string categories were retained in the prepared dataset.

**Feature selection:** The raw per-sensor temperature, light, and sound columns were highly correlated with each other and with their group averages, so they were dropped from the model-ready feature set in favor of aggregate features to reduce redundancy and multicollinearity. `S5_CO2`, `S5_CO2_Slope`, `S6_PIR`, and `S7_PIR` were retained as-is.

**Scaling:** `StandardScaler` was fit on the training set only and applied to both training and test sets, avoiding leakage of test-set statistics into the scaling parameters. Scaling primarily benefits distance-based and linear models; tree-based models do not require it.
